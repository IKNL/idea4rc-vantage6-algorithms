# New RAVEN Workspace
*New study in vantage6*

```python
import requests
```

# Set the collaboration ID:
#
#   - 2: Test collaboration (with IKNL and UPM)
#   - 3: IDEA4RC collaboration
#
COLLABORATION_ID = 2

```python
with open("token.txt", "r") as f:
    token = f.read().strip()
headers = {
    "Authorization": token
}
```

```python
# Set the organization IDs that are part of this workspace. These should be the IDs of
# the vantage6 organizations:
#
#   1	- root
#   2	- ENG
#   3	- UPM
#   4	- INT
#   5	- UKE
#   6	- CLB
#   7	- FPNS
#
# NOTE 1:
# Organization should be part of the collaboration. I recommend logging into the
# vantage6 UI and check the organizations in the collaboration. Alternatively,
# you can use the API to get the organizations:
response = requests.get(
    f"https://vantage6-core.orchestrator.idea.lst.tfo.upm.es/server/organization?collaboration_id={COLLABORATION_ID}",
    headers=headers
)
ORGANIZATION_IDS = [org["id"] for org in response.json()["data"]]
ORGANIZATION_IDS
# NOTE 2:
# These organization IDs should have been stored in the RAVEN database
# (`v6_organization`).
```

```python
# To create a new study we need the organizations ids (the internal ids in vantage6)
# that are included in this workspace. The name of the study needs to be unique.
response = requests.post(
    "https://vantage6-core.orchestrator.idea.lst.tfo.upm.es/server/study",
    headers=headers,
    json={
        # The collaboration id is the vantage6 id of the collaboration. This is
        # is the same for all workspaces. I used 1 now, but this can change when we
        # are still developing the platform.
        "collaboration_id": COLLABORATION_ID,
        # NOTE --- CHANGE THE NAME OF THE STUDY  ---
        # The name of the study needs to be unique. I guess the name of the workspace
        # is also unique, so we can use that. You could consider using a UUID.
        "name": "UPM Test Study 0",
        # The organization ids are the internal ids of the organizations in vantage6.
        "organization_ids": ORGANIZATION_IDS,
    }
)
response.json()
# In the case that:
#
# - The name is not unique
# - The collaboration id is not valid (non existing)
# - The organization ids are not valid (non existing)
#
# The API will return a 4xx error with a message. It will be of the following format:
# {
#     "msg": "Error message",
# }
```

```python
# This study ID should be stored in the RAVEN database in the workspace table
# (`v6_workspace`).
STUDY_ID = response.json()["id"]
STUDY_ID
```
