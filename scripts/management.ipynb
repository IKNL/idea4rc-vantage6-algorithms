```python
import json
with open("./pts_lists.json", "r") as f:
    pts_list = json.load(f)
```

```python
pts_list.keys()
```

```python
patient_ids_rps = pts_list["pts_RPS"]
patient_ids_pelvis = pts_list["pts_Pelvis"]
patient_ids_pelvis_rps = pts_list["pts_RPS_Pelvis"]
```

```python
client.dataframe.create(
    name="RPS",
    label="omop",
    image="harbor2.vantage6.ai/idea4rc/sessions:latest",
    method="create_cohort",
    arguments={
        "patient_ids": patient_ids_rps,
        "features": "sarcoma"
    },
    session=2
)
```

```python
client.dataframe.create(
    name="Pelvis",
    label="omop",
    image="harbor2.vantage6.ai/idea4rc/sessions:latest",
    method="create_cohort",
    arguments={
        "patient_ids": patient_ids_pelvis,
        "features": "sarcoma"
    },
    session=2
)
```
